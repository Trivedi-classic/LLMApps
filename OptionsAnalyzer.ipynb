# Request Options Data
Request options chain data for a specified underlying asset, including strikes, expirations, and types.

```python
# Request options chain data for a specified underlying asset, including strikes, expirations, and types

# Define the underlying asset (e.g., AAPL stock)
underlying = Stock('NFLX', 'SMART', 'USD')

# Qualify the underlying contract to get its conId
ib.qualifyContracts(underlying)
print(f"Qualified contract for {underlying.symbol} with conId: {underlying.conId}")
```

```python
# # Request the options chain for the underlying asset
# options_chains = ib.reqSecDefOptParams(underlying.symbol, underlying.secType, underlying.exchange,underlying.conId)
nflx = Stock("NFLX", "SMART", "USD")
ib.qualifyContracts(nflx)  # Qualify the contract first
[ticker] = ib.reqTickers(nflx)
nflxValue = ticker.marketPrice()
today = datetime.now().date()

chains = ib.reqSecDefOptParams(nflx.symbol, "", nflx.secType, nflx.conId)
chain = next((c for c in chains if c.tradingClass == "NFLX" and c.exchange == "SMART"), None)
if chain is None:
    print(f"No options chain found. Available chains: {chains}")
    raise ValueError("Unable to find NVDA options chain")

strikes = [
    strike for strike in chain.strikes
    if strike % 5 == 0
    and nflxValue - 50 < strike < nflxValue + 50
]
expirations = sorted(
    exp for exp in chain.expirations 
    if 20 <= (datetime.strptime(exp, '%Y%m%d').date() - today).days <= 31
)
rights = ["C", "P"]
```

# Extract and Display Options Data
Extract the options data from the response and display it in a structured format, such as a DataFrame.

```python
contracts = [
    Option("NFLX", expiration, strike, right, "SMART", tradingClass="NFLX")
    for right in rights
    for expiration in expirations
    for strike in strikes
]

contracts = ib.qualifyContracts(*contracts)
tickers = ib.reqTickers(*contracts)

# ib.sleep(1) # Ensure both Greeks and bid/ask have streamed in

filtered_tickers = [
    t for t in tickers 
    if t.modelGreeks and 0.15 <= abs(t.modelGreeks.delta) <= 0.30
]

contractData = [
    (
        t.contract.lastTradeDateOrContractMonth, 
        t.contract.strike, 
        t.contract.right,
        t.time,
        nflxValue,
        t.close,
        t.bid,   # Added Bid
        t.ask,   # Added Ask
        t.modelGreeks.delta
    )
    for t in filtered_tickers #filtered_tickers
]

fields = ["expiration", "strike", "right", "time", "undPrice", "close", "bid", "ask", "delta"]
```

```python
df = pd.DataFrame(contractData, columns=fields)

if not df.empty:
    # Convert to datetime and calculate DTE
    df['expiration'] = pd.to_datetime(df['expiration'], format='%Y%m%d')
    today_ts = pd.to_datetime(today)
    
    # This syntax is robust against the NumPy/Timestamp error
    df['DTE'] = (df['expiration'] - today_ts).dt.days
    
    # Sort and format
    df = df.sort_values(by=['expiration', 'strike'], ascending=True)
    df['expiration'] = df['expiration'].dt.strftime('%Y-%m-%d')
    
    print(df)
else:
    print("No contracts found matching the Delta and DTE criteria.")
```

```python
# Analyse the df options DataFrame to identify potential trading opportunities based on specific criteria such as Delta and Days to Expiration (DTE).

# 1. Initialize the OpenAI client
# Ensure your API key is set as an environment variable 'OPENAI_API_KEY'

import openai
from openai import OpenAI
openai_api_key = os.getenv('OPENAI_API_KEY')
print("OpenAI API Key:", openai_api_key)
openai = OpenAI()

# 2. Convert your DataFrame to a Markdown string for the LLM
# Note: Requires the 'tabulate' library (pip install tabulate)
data_markdown = df.to_markdown(index=False)

# 3. Define the LLM instructions and the user request
system_prompt = (
    "You are an expert hedge fund manager with focus on generating options income. "
    "You are given the dataframe of options data for a particular stock ticker and will "
    "identify which contract to trade and identify based on the premium whether it is a "
    "good idea to do covered call or cash secured put strategy. Note, that the given "
    "stock ticker is already in neutral to bullish trend."
)

user_prompt = f"Here is the current NVDA options data:\n\n{data_markdown}"

# --- STEP 4: Initial Analysis (The Strategist) ---
response_1 = openai.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ],
    temperature=0.2
)
initial_draft = response_1.choices[0].message.content

# --- STEP 5: Reflection Step (The Critic) ---
reflection_prompt = (
    f"You are a risk-focused hedge fund auditor. Review the following trade proposal:\n\n"
    f"{initial_draft}\n\n"
    "Critique this proposal by considering:\n"
    "1. Does the premium justify the risk of assignment (for CSP) or capped upside (for CC)?\n"
    "2. Are there any earnings events or macro factors the proposal might have ignored?\n"
    "3. Is the Delta chosen too aggressive given the 'neutral to bullish' context?\n"
    "Identify any flaws and suggest a more optimal adjustment if needed."
)

response_2 = openai.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "You are a critical risk auditor."},
        {"role": "user", "content": reflection_prompt}
    ],
    temperature=0.1 # Very low for high consistency
)
reflection_critique = response_2.choices[0].message.content

# --- STEP 6: Final Refinement (The Refiner) ---
final_prompt = (
    f"Based on your initial plan:\n{initial_draft}\n\n"
    f"And the following critique:\n{reflection_critique}\n\n"
    "Provide a final, optimized recommendation for the given ticker options trade."
)

response_3 = openai.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": final_prompt}
    ],
    temperature=0.2
)

# --- STEP 7: Print the Final Outcome ---
print("--- FINAL OPTIMIZED RECOMMENDATION ---")
print(response_3.choices[0].message.content)

```
